## 1. Définition générale

L'apprentissage par renforcement (RL) est une branche de l'intelligence artificielle où un **agent** apprend à prendre des décisions en interagissant avec un **environnement**. L'objectif est de maximiser une **récompense cumulative** au fil du temps.

L'agent apprend par essai-erreur : il reçoit des récompenses (ou des punitions) selon la qualité de ses actions.

À chaque pas de temps, la boucle d'interaction est la suivante : l'agent observe l'état de l'environnement, choisit une action, l'environnement évolue et renvoie une nouvelle observation et une récompense.

## 2. Les éléments fondamentaux du RL : le MDP

Le RL repose sur un modèle formel appelé **Processus de Décision de Markov** (MDP), composé de cinq éléments :

| Élément | Description | Exemple (CartPole) |
|---|---|---|
| S : états | Situations possibles de l'agent | Position et vitesse du chariot, angle et vitesse angulaire de la tige |
| A : actions | Ce que l'agent peut faire | Déplacer le chariot à gauche ou à droite |
| R : récompense | Score reçu après chaque action | +1 si la tige reste droite |
| P : transition | Probabilité de passer d'un état à un autre après une action | Mécanique physique du système |
| γ : facteur d'actualisation | Pondère l'importance des récompenses futures (entre 0 et 1) | 0.9 : les récompenses futures sont valorisées presque autant que les immédiates |

Le retour (récompense cumulée actualisée) à l'instant t s'écrit :

$$G_t = \sum_{k=0}^{\infty} \gamma^k \, r_{t+k+1}$$

Notions associées :

- **Politique** π(a | s) : probabilité de choisir l'action a dans l'état s.
- **Fonction de valeur** V^π(s) : retour espéré en partant de s et en suivant π.
- **Fonction action-valeur** Q^π(s, a) : retour espéré en choisissant a dans s puis en suivant π.
- **Exploration / exploitation** : l'agent doit essayer de nouvelles actions pour découvrir de meilleures stratégies (exploration) tout en utilisant ce qu'il sait déjà être efficace (exploitation).

## 3. Les grandes familles d'algorithmes RL

### Méthodes de valeurs (value-based)

On apprend quelle est la meilleure action dans chaque état.

- Q-Learning
- Deep Q-Network (DQN)

### Méthodes de politique (policy-based)

On apprend directement la politique π sans passer par une table de valeurs.

- REINFORCE
- PPO (Proximal Policy Optimization), très utilisé dans les outils modernes comme RLlib ou Stable-Baselines3

### Méthodes acteur-critique (actor-critic)

Elles combinent les deux approches :

- **Acteur** : choisit les actions (politique)
- **Critique** : estime la valeur des actions

Exemples : A2C, A3C, PPO, DDPG, SAC.

## 4. Fonction de récompense et apprentissage

L'agent apprend grâce à la fonction de récompense :

- si l'action améliore la situation, la récompense est positive ;
- si elle l'aggrave, la récompense est négative.

C'est cette fonction de feedback qui guide tout l'apprentissage.

## 5. Exemple simple : CartPole

L'agent doit maintenir une tige droite sur un chariot mobile.

- **État** : position et vitesse du chariot, angle et vitesse angulaire de la tige
- **Actions** : 0 = pousser à gauche, 1 = pousser à droite
- **Récompense** : +1 à chaque pas où la tige reste droite
- **Objectif** : maximiser la durée avant que la tige tombe

Fin d'un épisode dans CartPole-v1 :

- `terminated = True` : l'angle de la tige dépasse environ 12° ou le chariot sort de la zone (|x| > 2.4) ;
- `truncated = True` : l'épisode atteint 500 pas.

La récompense valant +1 par pas, la récompense totale d'un épisode est égale à sa durée en pas. Le maximum est 500.

## 8. Exemple 1 : entraînement PPO puis un épisode de test

Étapes du code :

1. `gym.make("CartPole-v1")` crée l'environnement. L'entraînement se fait sans affichage, car `render_mode="human"` pendant `learn` ralentit fortement l'apprentissage. Pour voir la simulation, mettre `RENDER = True` : l'affichage n'est alors activé que pendant le test.
2. `PPO("MlpPolicy", env)` crée un agent dont l'acteur et le critique sont des réseaux de neurones (2 couches de 64 neurones par défaut).
3. `model.learn(total_timesteps=10000)` entraîne l'agent. Avec `n_steps=2048` par défaut, cela fait environ 5 itérations (5 × 2048 = 10 240 pas) : collecte de 2048 transitions, mise à jour de la politique sur 10 époques, puis nouvelle collecte.
4. Le test exécute un épisode avec `deterministic=True` : l'agent choisit l'action la plus probable au lieu d'en tirer une au hasard.

```python
train_env = gym.make("CartPole-v1")
model = PPO("MlpPolicy", train_env, verbose=1, seed=SEED)

print("Entraînement en cours...")
model.learn(total_timesteps=TOTAL_TIMESTEPS)
print("Entraînement terminé !")
train_env.close()
```

```python
test_env = gym.make("CartPole-v1", render_mode="human" if RENDER else None)

obs, info = test_env.reset(seed=SEED)
done = False
total_reward = 0.0
steps = 0

while not done:
    action, _ = model.predict(obs, deterministic=True)
    obs, reward, terminated, truncated, info = test_env.step(action)
    done = terminated or truncated
    total_reward += reward
    steps += 1

test_env.close()

print(f"Récompense totale de l'épisode : {total_reward}")
print(f"Nombre de pas : {steps}")
print(f"Fin par chute ou sortie de piste : {terminated}")
print(f"Fin par limite de 500 pas : {truncated}")
```

## 9. Exemple 2 : cinq épisodes de test et courbe des récompenses

Différences avec l'exemple 1 :

- le test est répété sur 5 épisodes et chaque récompense totale est stockée dans une liste ;
- `matplotlib` trace la récompense totale en fonction du numéro d'épisode ;
- un agent aléatoire sert de référence pour mesurer le gain apporté par l'apprentissage.

```python
def run_episodes(env, policy, num_episodes, seed):
    rewards = []
    for ep in range(num_episodes):
        obs, info = env.reset(seed=seed + ep)
        done = False
        total_reward = 0.0
        while not done:
            action = policy(obs)
            obs, reward, terminated, truncated, info = env.step(action)
            done = terminated or truncated
            total_reward += reward
        rewards.append(total_reward)
        print(f"Épisode {ep + 1} - Récompense : {total_reward}")
    return rewards


eval_env = gym.make("CartPole-v1", render_mode="human" if RENDER else None)

print("Agent PPO entraîné")
ppo_rewards = run_episodes(
    eval_env,
    lambda obs: model.predict(obs, deterministic=True)[0],
    NUM_EPISODES,
    SEED,
)

print("Agent aléatoire (référence)")
random_rewards = run_episodes(
    eval_env,
    lambda obs: eval_env.action_space.sample(),
    NUM_EPISODES,
    SEED,
)

eval_env.close()

print(f"PPO : moyenne = {np.mean(ppo_rewards):.1f}, écart-type = {np.std(ppo_rewards):.1f}")
print(f"Aléatoire : moyenne = {np.mean(random_rewards):.1f}, écart-type = {np.std(random_rewards):.1f}")
```

```python
episodes = list(range(1, NUM_EPISODES + 1))

plt.figure(figsize=(8, 4))
plt.plot(episodes, ppo_rewards, marker="o", label="PPO entraîné")
plt.plot(episodes, random_rewards, marker="s", linestyle="--", label="Agent aléatoire")
plt.axhline(500, color="gray", linestyle=":", label="Maximum (500)")
plt.title("Récompense totale par épisode")
plt.xlabel("Épisode")
plt.ylabel("Récompense")
plt.xticks(episodes)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig("recompenses_par_episode.png", dpi=150)
plt.show()
```

L'entropie diminue lentement : la politique devient moins aléatoire (l'agent passe progressivement de l'exploration à l'exploitation).

**Test.** La récompense totale de l'épisode est égale au nombre de pas pendant lesquels la tige est restée debout :

- proche de 500 : la politique stabilise la tige pendant toute la durée autorisée, l'épisode se termine par `truncated` ;
- nettement inférieure (par exemple 100 à 200) : la politique est meilleure que le hasard mais pas encore stable, l'épisode se termine par `terminated` ;
- proche de 10 à 30 : la politique est au niveau du hasard, l'entraînement est insuffisant.

Limite : un seul épisode ne permet pas de conclure, car le résultat dépend de l'état initial tiré au hasard et de la graine d'entraînement.

## 11. Évaluation approfondie et courbe d'apprentissage

Pour juger correctement un agent, on évalue sur davantage d'épisodes avec `evaluate_policy`, et on enregistre les récompenses pendant l'entraînement avec `Monitor` pour tracer une vraie courbe d'apprentissage. Ici l'agent est entraîné pendant 50 000 pas et comparé à l'agent de 10 000 pas.

```python
monitored_env = Monitor(gym.make("CartPole-v1"))
long_model = PPO("MlpPolicy", monitored_env, verbose=0, seed=SEED)
long_model.learn(total_timesteps=50000)

episode_rewards = np.array(monitored_env.get_episode_rewards())
window = 20
moving_average = np.convolve(episode_rewards, np.ones(window) / window, mode="valid")

plt.figure(figsize=(9, 4))
plt.plot(episode_rewards, alpha=0.35, label="Récompense par épisode")
plt.plot(range(window - 1, len(episode_rewards)), moving_average, label=f"Moyenne glissante ({window})")
plt.axhline(500, color="gray", linestyle=":", label="Maximum (500)")
plt.title("Courbe d'apprentissage de PPO sur CartPole-v1")
plt.xlabel("Épisode d'entraînement")
plt.ylabel("Récompense")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig("courbe_apprentissage.png", dpi=150)
plt.show()

monitored_env.close()
```

```python
final_env = gym.make("CartPole-v1")

mean_10k, std_10k = evaluate_policy(model, final_env, n_eval_episodes=20, deterministic=True)
mean_50k, std_50k = evaluate_policy(long_model, final_env, n_eval_episodes=20, deterministic=True)

final_env.close()

results = pd.DataFrame(
    {
        "Entraînement (pas)": [TOTAL_TIMESTEPS, 50000],
        "Récompense moyenne": [mean_10k, mean_50k],
        "Écart-type": [std_10k, std_50k],
    }
)
print(results.to_string(index=False))

plt.figure(figsize=(6, 4))
plt.bar(["10 000 pas", "50 000 pas"], [mean_10k, mean_50k], yerr=[std_10k, std_50k], capsize=6)
plt.axhline(500, color="gray", linestyle=":")
plt.title("Récompense moyenne sur 20 épisodes de test")
plt.ylabel("Récompense")
plt.tight_layout()
plt.savefig("comparaison_entrainements.png", dpi=150)
plt.show()
```

### Interprétation

- La courbe d'apprentissage monte au fil des épisodes d'entraînement : c'est elle qui montre la progression de l'agent, contrairement à la courbe de test de la section 9.
- Une moyenne glissante lisse le bruit dû à l'exploration et aux états initiaux aléatoires.
- Avec 50 000 pas, PPO atteint en général une récompense proche ou égale à 500 sur CartPole, avec un écart-type faible. Avec 10 000 pas, la moyenne est plus basse et la dispersion plus forte.
- Un écart-type élevé signale une politique instable. Un écart-type nul à 500 signale une politique qui a résolu la tâche.

## 12. Exploration / exploitation et sensibilité à la graine

Les résultats d'un algorithme de RL varient d'une graine à l'autre. Pour le montrer, on entraîne PPO avec plusieurs graines et on compare les récompenses moyennes obtenues.

```python
seeds = [0, 1, 2, 3, 4]
seed_means = []

for s in seeds:
    seed_env = gym.make("CartPole-v1")
    seed_model = PPO("MlpPolicy", seed_env, verbose=0, seed=s)
    seed_model.learn(total_timesteps=TOTAL_TIMESTEPS)
    mean_reward, _ = evaluate_policy(seed_model, seed_env, n_eval_episodes=10, deterministic=True)
    seed_means.append(mean_reward)
    seed_env.close()
    print(f"Graine {s} : récompense moyenne = {mean_reward:.1f}")

plt.figure(figsize=(7, 4))
plt.bar([str(s) for s in seeds], seed_means)
plt.axhline(np.mean(seed_means), color="red", linestyle="--", label=f"Moyenne = {np.mean(seed_means):.1f}")
plt.title("PPO (10 000 pas) selon la graine aléatoire")
plt.xlabel("Graine")
plt.ylabel("Récompense moyenne (10 épisodes)")
plt.legend()
plt.tight_layout()
plt.savefig("sensibilite_graine.png", dpi=150)
plt.show()
```

La dispersion entre graines confirme qu'un seul entraînement ne suffit pas pour conclure : il faut répéter l'expérience et rapporter la moyenne et l'écart-type.

## 13. Entraînement distribué avec RLlib (Ray)

RLlib exécute la collecte d'expérience en parallèle dans plusieurs processus (*rollout workers*), chacun possédant sa copie de l'environnement. Le *learner* centralise les transitions et met à jour la politique. Ceci permet de passer à l'échelle sur plusieurs cœurs ou plusieurs machines.

Cette section nécessite `ray[rllib]==2.9.1` (le paquet `ray[default]` seul n'inclut pas RLlib). Elle est désactivée par défaut : mettre `RUN_RLLIB = True` dans la cellule de configuration pour l'exécuter. Ray fonctionne mieux sous Linux ou macOS ; sous Windows, la prise en charge est plus limitée.

```python
if RUN_RLLIB:
    %pip install -q "ray[rllib]==2.9.1"
```

```python
if RUN_RLLIB:
    import ray
    from ray.rllib.algorithms.ppo import PPOConfig

    ray.init(ignore_reinit_error=True)

    config = (
        PPOConfig()
        .environment("CartPole-v1")
        .framework("torch")
        .rollouts(num_rollout_workers=2)
        .training(train_batch_size=4000)
    )

    algo = config.build()
    rllib_rewards = []

    for iteration in range(10):
        result = algo.train()
        rllib_rewards.append(result["episode_reward_mean"])
        print(f"Itération {iteration + 1} - récompense moyenne : {result['episode_reward_mean']}")

    algo.stop()
    ray.shutdown()

    plt.figure(figsize=(8, 4))
    plt.plot(range(1, len(rllib_rewards) + 1), rllib_rewards, marker="o")
    plt.axhline(500, color="gray", linestyle=":")
    plt.title("RLlib PPO sur CartPole-v1 (2 workers)")
    plt.xlabel("Itération d'entraînement")
    plt.ylabel("Récompense moyenne")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig("rllib_apprentissage.png", dpi=150)
    plt.show()
```
